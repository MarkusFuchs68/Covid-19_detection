```python
def load_and_make_resized_image_array(image_path):
    # Load and convert it to numpy array of shape (224,224,3) and pixel-value-range 0-255
    img = image.load_img(image_path, target_size=(224, 224))  # Resize if needed
    return image.img_to_array(img)
```

##### Model selection

```python
# This is the model, which we run experiments here
# you can take any model, you have trained before and load its saved params
model = keras.models.load_model('vgg16_2-classes.keras')
```

##### Masked vs. Unmasked

> Gain insights into what the model does with masked or unmasked images

```python
# Take a random COVID image
img_name = 'COVID-12.png'

# Take a masked image
masked_img_path = "../dataset/covid/" + img_name
unmasked_img_path = '../COVID-19_Radiography_Dataset/COVID/images/' + img_name
```

>> Masked feature maps

We can see, that the model extracts especially the edges of the mask, which is the least information the model should evaluate. Hence, we consider that masking actually confuses the model more than unmasking.

```python
show_feature_maps(model, load_and_make_resized_image_array(masked_img_path))
```

>> Unmasked feature maps

We can see, that the model extracts more details of the lungs itself. 

Hence this might result in better performance. Also if this would be the case, we can use the model without prior masking, which makes its application easier.

```python
show_feature_maps(model, load_and_make_resized_image_array(unmasked_img_path))
```

>> We can see, that on the unmasked feature maps, there is more detail about the lung and not about the edges of the lung. Let's see, if there is a difference in prediction.

```python
# Load, resize, adapt shape, normalize (preprocess) for model
masked_image = load_and_make_resized_image_array(masked_img_path)
unmasked_image = load_and_make_resized_image_array(unmasked_img_path)

# Add an extra dimension to the image to make it compatible with the batch (shape: (1, H, W, C))
masked_image_batch = tf.expand_dims(masked_image, axis=0)
unmasked_image_batch = tf.expand_dims(unmasked_image, axis=0)

# Normalize (optional, depends on model)
masked_image_batch = masked_image_batch / 255.0  # Normalize to [0,1]
unmasked_image_batch = unmasked_image_batch / 255.0  # Normalize to [0,1]
```

```python
# Now predict
pred_masked = model.predict(masked_image_batch)
pred_unmasked = model.predict(unmasked_image_batch)

# We remember from our notebook, where the model was trained, 
# that[0] corresponds to the normal and [1] corresponds to the covid class
print('Masked prediction:', pred_masked, '->', np.argmax(pred_masked))
print('Unmasked prediction', pred_unmasked, '->', np.argmax(pred_unmasked))
```
